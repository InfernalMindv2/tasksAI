```python
import sys
board  = {1:' ' , 2: ' ' , 3: ' ',
         4: ' ' , 5: ' ' , 6: ' ',
         7: ' ' , 8: ' ', 9: ' '}
#This is the initial board before running !! 
symbol = 'O'
test = symbol
mxdepth=0
```

```python
#This is function is used for output the current state of our game !
def printboard(board):
    print(board[1]+'|'+board[2]+'|'+board[3])
    print('-+-+-')
    print(board[4]+'|'+board[5]+'|'+board[6])
    print('-+-+-')
    print(board[7]+'|'+board[8]+'|'+board[9])
    print('-+-+-')
    print('\n')
```

```python
#This function is to check if our cell is free or not ! 
def IsSpaceFree(number,board):
    if board[int(number)] == ' ':
        return True
    else:
        return False
```

```python
#This function is to insert the Symbol based on the current turn
#If the number doesnt qualify the requirements , we go in infinity loop and trying to attempt a vaild position
def InsertAtPosition(number,board):
    if InputChar(number) and IsSpaceFree(number,board):
        board[int(number)]=symbol
        print("Successful Insertion !!")
        return
    else:
        while True:
            print("Invalid!!")
            print("Enter the position again!!")
            printboard(board)
            positionagain = input()
            if InputChar(positionagain) and IsSpaceFree(positionagain,board):
                InsertAtPosition(positionagain,board)
                break
```

```python
#We create this function in order to check the input is digit and between 1 and 9 inclusively .
def InputChar(position):
    if position.isdigit() and int(position)>=1 and int(position)<=9:
        return True
    else:
        return False
```

```python
#To determine whether th turn is X or O
def ChangePlay():
    global symbol
    if(symbol=='X'):
        symbol='O'
    else:
        symbol='X'
```

```python
#This function that display the current state to play at empty cell.
#We first ask the play to insert the position and going to function InsertAtPosition to do all of requied procedures.
def YourTurn():
    printboard(board)
    print("Enter your position : ")
    number = input("")
    InsertAtPosition(number,board)
```

```python
def IsDraw():
    for i in board.values():
        if i == ' ':
            return False
    return True
```

```python
#This function has all valid cases of winning :D
def IsWin(symbol):
    if ( (board[1]==symbol and board[2]==symbol and board[3]==symbol) or
    (board[4]==symbol and board[5]==symbol and board[6]==symbol) or
    (board[7]==symbol and board[8]==symbol and board[9]==symbol) or
    (board[1]==symbol and board[4]==symbol and board[7]==symbol) or
    (board[2]==symbol and board[5]==symbol and board[8]==symbol) or
    (board[3]==symbol and board[6]==symbol and board[9]==symbol) or
    (board[1]==symbol and board[5]==symbol and board[9]==symbol) or
    (board[3]==symbol and board[5]==symbol and board[7]==symbol) ):
        return True
    else:
        return False
```

```python
#This function is our bot , we simple try to play in all empty cells and go recursively to all best scenarios of that cell
def YourBot():
    bestScore = -1000
    bestMove = -1
    for key in board.keys():
        test='X'
        if board[key] == ' ':
            board[key]='X'
            score = miniMax(board,1,False)
            board[key] = ' '
            if score > bestScore:
                bestScore = score
                bestMove = key
    InsertAtPosition(str(bestMove),board)
    return
```

```python

def miniMax(board,depth,isMaxOrMin):
    global mxdepth
    if IsWin('X'):
        return 1000-depth
    elif IsWin('O'):
        return -1000+depth
    elif IsDraw() or mxdepth<=depth:
        return 0
    if isMaxOrMin:
        bestScore = -1000
        for key in board.keys():
            if board[key] == ' ':
                board[key]='X'
                score = miniMax(board,depth+1,False)
                board[key] = ' '
                if score > bestScore:
                    bestScore = score
        return bestScore
    else:
        bestScore = 1000
        for key in board.keys():
            if board[key] == ' ':
                board[key]='O'
                score = miniMax(board,depth+1,True)
                board[key] = ' '
                if score < bestScore:
                    bestScore = score
        return bestScore
```

```python
#Our main program.
#Semi console's gui to understand how to play
#We first start the level of difficulty by entering 1 ,2,3
#After that we every iteration of loop , we check if it is draw , win for player or bot.
#mxdepth is like how many levels in our tree will will my code goes into ?
#Like if i have 6 empty cells in my current state and i have mxdepth = 3 .Then i should play the best minimax atmost  3 movements. 
while True:
    te=input("Hello , please enter the level of game : 1)Easy   2) Medium    3) Hard")
    if (te.isdigit() and te=="1"):
        mxdepth=1
        break
    elif (te.isdigit() and te=="2"):
        mxdepth = 4
        break
    elif (te.isdigit() and te=="3"):
        mxdepth = 20
        break
    else:
        print("Invalid!!")
        
while True:
    if IsDraw():
        printboard(board)
        print("Draw!")
        break
    elif (IsWin('O')):
        printboard(board)
        print("You win!!")
        break
    elif IsWin('X'):
        printboard(board)
        print("Bot wins!!")
        break
    if symbol=='X':
        YourBot()
    else:
        YourTurn()
    ChangePlay()
input("Press any to quit !")
```
